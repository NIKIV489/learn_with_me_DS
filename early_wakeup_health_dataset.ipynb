```python
"""looks like:

[
 [ax1, ax2],
 [ax3, ax4],
 [ax5, ax6],
 [ax7, ax8]
]

After flatten:

[
 ax1,
 ax2,
 ax3,
 ax4,
 ax5,
 ax6,
 ax7,
 ax8 helps loop through each axis in a 2D array
]"""
```

```python
import seaborn as sns

cols = [
    'Sleep_Time',
    'Wake_Up_Time',
    'Exercise_Type',
    'Workout_Intensity',
    'Smoking_Status',
    'Alcohol_Consumption',
    'Diabetes_Risk',
    'Wellness_Category'
]

fig, axes = plt.subplots(4, 2, figsize=(18, 16))
axes = axes.flatten()

for i, col in enumerate(cols):
    sns.countplot(
        data=a,
        x=col,
        order=a[col].value_counts().index,
        ax=axes[i]
    )
    axes[i].set_title(col)
    axes[i].tick_params(axis='x', rotation=90)

plt.tight_layout()
plt.show()
```
