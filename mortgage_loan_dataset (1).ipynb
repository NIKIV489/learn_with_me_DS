```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

x = a["Age"]
y = a["Annual Income (USD)"]
z = a["Credit Score"]

fig = plt.figure(figsize=(10,7))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(x,y,z)

ax.set_xlabel("Age")
ax.set_ylabel("Annual Income")
ax.set_zlabel("Credit Score")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# take sample because 49990 rows is heavy
b = a.sample(49990)

X = b["Age"].values
Y = b["Annual Income (USD)"].values
Z = b["Credit Score"].values

fig = plt.figure(figsize=(10,7))

#111 = 1 row, 1 column, 1st graph
ax = fig.add_subplot(111, projection="3d")

# wave style using your data
ax.plot_trisurf(X,Y,Z)

ax.set_xlabel("Age")
ax.set_ylabel("Annual Income")
ax.set_zlabel("Credit Score")

plt.show()
a.shape
```
