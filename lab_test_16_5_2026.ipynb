```python
#college wants to analyze student marks of 500 students and identify toppers, average performers, and failed students using Python and NumPy.
import numpy as np
a=np.random.randint(1,101,500)
a
topper=[]
average=[]
failed=[]
def marks(num):
  for i in num:
    if i>=90:
      topper.append(int(i))
    elif 45<=i<90:
      average.append(int(i))
    elif i<45:
      failed.append(int(i))
marks(a)
print("The topper mark : ", topper)
print("The Average mark : ",average)
print("the failed mark : ",failed)
```
