```python
num_cols = a.select_dtypes(include="number").columns
cat_cols = a.select_dtypes(include="object").columns

a[num_cols] = a[num_cols].fillna(a[num_cols].mean())
a[cat_cols] = a[cat_cols].fillna(a[cat_cols].mode().iloc[0])
```

```python
y = a["RainTomorrow"]          # Save target first
x = a.drop("RainTomorrow", axis=1)
```

```python
x = pd.get_dummies(x, drop_first=True)
```

```python
from sklearn.model_selection import train_test_split

```

```python
x_test,x_train,y_test,y_train=train_test_split(x,y,train_size=0.7,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
aa=LogisticRegression()
```

```python
x.dtypes.value_counts()
```

```python
aa.fit(x_train, y_train)
```

```python
y_pred = aa.predict(x_test)
```
