```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
plt.plot([81,83,85,73,85,91],marker='d',linestyle='dashdot',ms=7,mec="black",mfc="yellow",linewidth=2)
plt.legend()
plt.show()
```

```python
#bar chart
subjects=["english","tamil","maths","science","social"]
mark=[89,99,89,91,99]
plt.bar(subjects,mark,color=['yellow','violet','skyblue','gray',"brown"])
plt.show()
```

```python
#pie chart
products=['a','b','c','d']
quantity=[4200,4500,4000,4000]
plt.pie(quantity,labels=products,colors=['yellow','violet','skyblue','gray'],autopct='%1.1f%%')
plt.show()
```

```python
#histogram
marks=[35,36,50,75,92,98,93]
plt.hist(marks,color='violet')
plt.show()
```

```python
import pandas as pd
t=pd.read_csv("/content/Titanic-Dataset.csv")
t
```

```python
plt.hist(t['Age'],color='LIGHTGREEN')
plt.show()
```

```python
plt.bar(t['Sex'],t['Survived'],color=['PINK','gray'])
plt.xlabel("sex")
plt.ylabel("survived ")
plt.title("gender vs survived")
plt.show()
```

```python
#sscatter plot
plt.scatter(t['Age'],t['Fare'])
plt.show()
```

```python
print(t['Age'].isnull().sum())
print(t['Age'].describe())
```

```python
# Box plot of Age after handling missing values
t.boxplot(column='Age', by='Sex')
plt.title('Box Plot of Age by Sex (Missing Values Filled)')
plt.ylabel('Age')
plt.show()
```

```python
#area chart
plt.fill_between(t['Age'],t['Fare'])
plt.show()
```

```python
import re

text = "hello planet"

print(re.findall("[a-m]", text))

text2 = "My number is 123"
print(re.findall(r"\d", text2))


print(re.findall("he..o", "hello"))

print(re.findall("^hello", "hello planet"))


print(re.findall("planet$", "hello planet"))


print(re.findall("he.*o", "hello"))

print(re.findall("he.+o", "heeeello"))


print(re.findall("he.?o", "hello"))
print(re.findall("he.?o", "heo"))


print(re.findall("he.{2}o", "hello"))

text3 = "falls stays"
print(re.findall("falls|stays", text3))


match = re.search(r"(\d+)-(\d+)", "123-456")
if match:
    print(match.group(1))
    print(match.group(2))
```
