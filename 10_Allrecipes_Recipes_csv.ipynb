```python
import pandas as pd
a1=pd.read_csv("/content/10_Allrecipes_Recipes.csv")
```

```python
a1.info()
```

```python
a1.isnull().sum()
```

```python

obj=["ingredients"]
a1[obj]=a1[obj].fillna("Unknown")
```

```python
num_cols = [
    "calories",
    "fat",
    "carbs",
    "protein",
    "avg_rating",
    "total_ratings",
    "reviews",
    "servings"
]

for col in num_cols:
    a1[col] = a1[col].fillna(a1[col].mean())
```
